# Scentence 임베딩 정확도 테스트 결과 시각화

이 노트북은 임베딩 테스트 결과를 시각화합니다.

```python
# 필요한 패키지 설치
# 아래 명령어를 터미널에서 실행하거나, 주석을 해제하여 셀에서 실행하세요
# !pip install pandas matplotlib seaborn numpy

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from datetime import datetime

# 한글 폰트 설정 (Windows)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 스타일 설정
sns.set_style("whitegrid")
plt.style.use('seaborn-v0_8-darkgrid')
```

## 1. 데이터 로드

CSV 파일 경로를 지정하세요:

```python
# CSV 파일 경로 (실제 파일명으로 변경하세요)
CSV_PATH = "../evaluation_results/embedding_test_details_20260205_115300.csv"

# 데이터 로드
df = pd.read_csv(CSV_PATH)

print(f"총 테스트 수: {len(df)}")
print(f"\n컬럼: {list(df.columns)}")
df.head()
```

## 2. 정확도 분포 시각화

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Top-K 정확도 막대 그래프
accuracies = [
    df['accuracy_top1'].mean() * 100,
    df['accuracy_top3'].mean() * 100,
    df['accuracy_top5'].mean() * 100,
    df['accuracy_top10'].mean() * 100
]
k_values = ['Top-1', 'Top-3', 'Top-5', 'Top-10']

axes[0, 0].bar(k_values, accuracies, color=['#e74c3c', '#3498db', '#2ecc71', '#f39c12'])
axes[0, 0].set_ylabel('정확도 (%)')
axes[0, 0].set_title('Top-K 정확도')
axes[0, 0].set_ylim(0, 100)
for i, v in enumerate(accuracies):
    axes[0, 0].text(i, v + 2, f'{v:.1f}%', ha='center', fontweight='bold')

# 순위 분포 히스토그램
ranks_filtered = df[df['rank'] <= 50]['rank']  # 50위 이내로 필터링
axes[0, 1].hist(ranks_filtered, bins=20, color='skyblue', edgecolor='black')
axes[0, 1].set_xlabel('순위')
axes[0, 1].set_ylabel('빈도')
axes[0, 1].set_title('순위 분포 (50위 이내)')
axes[0, 1].axvline(ranks_filtered.median(), color='red', linestyle='--', label=f'중간값: {ranks_filtered.median():.1f}')
axes[0, 1].legend()

# 정답 발견 여부 파이 차트
found_counts = df['found_in_top_k'].value_counts()
axes[1, 0].pie(found_counts, labels=['발견', '미발견'], autopct='%1.1f%%', 
               colors=['#2ecc71', '#e74c3c'], startangle=90)
axes[1, 0].set_title('Top-10 내 정답 발견률')

# 순위별 누적 정확도
rank_cumulative = []
for k in range(1, 21):
    acc = (df['rank'] <= k).mean() * 100
    rank_cumulative.append(acc)

axes[1, 1].plot(range(1, 21), rank_cumulative, marker='o', linewidth=2, markersize=6)
axes[1, 1].set_xlabel('Top-K')
axes[1, 1].set_ylabel('누적 정확도 (%)')
axes[1, 1].set_title('누적 정확도 곡선')
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].axhline(70, color='green', linestyle='--', alpha=0.5, label='70%')
axes[1, 1].axhline(50, color='orange', linestyle='--', alpha=0.5, label='50%')
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```

## 3. 유사도 분포 시각화

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# 정답 노트 유사도 분포
axes[0, 0].hist(df['target_similarity'], bins=30, color='lightgreen', edgecolor='black', alpha=0.7)
axes[0, 0].set_xlabel('코사인 유사도')
axes[0, 0].set_ylabel('빈도')
axes[0, 0].set_title('정답 노트 유사도 분포')
axes[0, 0].axvline(df['target_similarity'].mean(), color='red', linestyle='--', 
                   label=f'평균: {df["target_similarity"].mean():.3f}')
axes[0, 0].legend()

# Top-1 노트 유사도 분포
axes[0, 1].hist(df['top1_similarity'], bins=30, color='lightcoral', edgecolor='black', alpha=0.7)
axes[0, 1].set_xlabel('코사인 유사도')
axes[0, 1].set_ylabel('빈도')
axes[0, 1].set_title('Top-1 노트 유사도 분포')
axes[0, 1].axvline(df['top1_similarity'].mean(), color='blue', linestyle='--',
                   label=f'평균: {df["top1_similarity"].mean():.3f}')
axes[0, 1].legend()

# 유사도 차이 분포
axes[1, 0].hist(df['similarity_gap'], bins=30, color='gold', edgecolor='black', alpha=0.7)
axes[1, 0].set_xlabel('유사도 차이 (Top-1 - 정답)')
axes[1, 0].set_ylabel('빈도')
axes[1, 0].set_title('유사도 차이 분포')
axes[1, 0].axvline(df['similarity_gap'].mean(), color='red', linestyle='--',
                   label=f'평균: {df["similarity_gap"].mean():.4f}')
axes[1, 0].axvline(0, color='green', linestyle='-', alpha=0.5, label='0 (동일)')
axes[1, 0].legend()

# 정답 vs Top-1 유사도 비교 (산점도)
sample_df = df.sample(min(500, len(df)))  # 500개 샘플
axes[1, 1].scatter(sample_df['target_similarity'], sample_df['top1_similarity'], 
                   alpha=0.5, s=20)
axes[1, 1].plot([0, 1], [0, 1], 'r--', label='y=x (동일)')
axes[1, 1].set_xlabel('정답 노트 유사도')
axes[1, 1].set_ylabel('Top-1 노트 유사도')
axes[1, 1].set_title('정답 vs Top-1 유사도 비교')
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```

## 4. 성능 지표 요약

```python
# 성능 지표 계산
metrics = {
    '총 테스트 수': len(df),
    'Top-1 정확도': f"{df['accuracy_top1'].mean()*100:.1f}%",
    'Top-3 정확도': f"{df['accuracy_top3'].mean()*100:.1f}%",
    'Top-5 정확도': f"{df['accuracy_top5'].mean()*100:.1f}%",
    'Top-10 정확도': f"{df['accuracy_top10'].mean()*100:.1f}%",
    'MRR': f"{df.apply(lambda x: 1/x['rank'] if x['rank'] <= 10 else 0, axis=1).mean():.4f}",
    '평균 순위': f"{df['rank'].mean():.2f}",
    '중간 순위': f"{df['rank'].median():.1f}",
    '평균 정답 유사도': f"{df['target_similarity'].mean():.4f}",
    '평균 Top-1 유사도': f"{df['top1_similarity'].mean():.4f}",
    '평균 유사도 차이': f"{df['similarity_gap'].mean():.4f}",
}

for key, value in metrics.items():
    print(f"{key}: {value}")
```

## 5. 상위/하위 성능 노트 분석

```python
# Top-1에 가장 많이 들어간 노트
top1_success = df[df['rank'] == 1]['target_note'].value_counts().head(10)

# 순위가 낮은(성능이 안좋은) 노트
low_rank = df[df['rank'] > 100][['target_note', 'korean_query', 'rank']].head(10)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Top-1 성공 노트
top1_success.plot(kind='barh', ax=axes[0], color='green', alpha=0.7)
axes[0].set_title('Top-1 성공 횟수 (상위 10개 노트)')
axes[0].set_xlabel('성공 횟수')

# 유사도가 높은 쿼리 예시
high_sim = df.nlargest(10, 'target_similarity')[['target_note', 'korean_query', 'target_similarity']]
axes[1].axis('off')
table_data = []
for _, row in high_sim.iterrows():
    table_data.append([row['target_note'][:15], row['korean_query'][:30], f"{row['target_similarity']:.3f}"])

table = axes[1].table(cellText=table_data,
                      colLabels=['노트', '쿼리', '유사도'],
                      cellLoc='left',
                      loc='center')
table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1.2, 1.5)
axes[1].set_title('높은 유사도를 보인 쿼리 (상위 10개)')

plt.tight_layout()
plt.show()

print("\n=== 순위가 낮은 노트 (개선 필요) ===")
print(low_rank.to_string(index=False))
```

## 6. 쿼리 길이 vs 성능 분석

```python
# 쿼리 길이 추가
df['query_length'] = df['korean_query'].str.len()

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# 쿼리 길이 vs 순위
axes[0].scatter(df['query_length'], df['rank'], alpha=0.5, s=20)
axes[0].set_xlabel('쿼리 길이 (문자 수)')
axes[0].set_ylabel('순위')
axes[0].set_title('쿼리 길이 vs 순위')
axes[0].set_ylim(0, 50)  # 50위 이내로 제한

# 쿼리 길이 구간별 정확도
df['length_bin'] = pd.cut(df['query_length'], bins=[0, 10, 20, 30, 50, 100], 
                          labels=['0-10', '10-20', '20-30', '30-50', '50+'])
length_acc = df.groupby('length_bin')['accuracy_top3'].mean() * 100

length_acc.plot(kind='bar', ax=axes[1], color='steelblue')
axes[1].set_xlabel('쿼리 길이 구간')
axes[1].set_ylabel('Top-3 정확도 (%)')
axes[1].set_title('쿼리 길이별 Top-3 정확도')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
